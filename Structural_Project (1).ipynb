## Problem Statement

Estimate the floor slab loads for a reinforced concrete slab.

Given:

- Slab thickness = 160 mm
- Concrete unit weight = 25 kN/m³
- Floor finishing = 1.2 kN/m²
- Ceiling/services = 0.6 kN/m²
- Live load = 3.0 kN/m²

Required:

1. Slab self-weight
2. Total dead load
3. Total service load
4. Factored load using:

Factored Load = 1.4D + 1.6L

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def slab_loads(thickness, gamma_c, finish, ceiling, live):

    self_weight = thickness * gamma_c

    dead_load = self_weight + finish + ceiling

    service_load = dead_load + live

    factored_load = 1.4 * dead_load + 1.6 * live

    return self_weight, dead_load, service_load, factored_load
```

```python
thickness = 0.16
gamma_c = 25

finish = 1.2
ceiling = 0.6
live = 3.0

sw, dead, service, factored = slab_loads(
    thickness,
    gamma_c,
    finish,
    ceiling,
    live
)

print(sw)
```

```python
results = pd.DataFrame({
    "Parameter":[
        "Self Weight",
        "Total Dead Load",
        "Total Service Load",
        "Factored Load"
    ],
    "Value (kN/m²)":[
        sw,
        dead,
        service,
        factored
    ]
})

results
```

```python
fig, ax = plt.subplots(figsize=(6,2))

ax.axis('off')

table = ax.table(
    cellText=results.values,
    colLabels=results.columns,
    loc='center'
)

plt.savefig("outputs/results_table.png",
            bbox_inches='tight')

plt.show()
```

```python
components = [
    sw,
    finish,
    ceiling,
    live
]

labels = [
    "Self Weight",
    "Finishing",
    "Ceiling",
    "Live Load"
]

plt.figure(figsize=(8,5))

plt.bar(labels, components)

plt.ylabel("Load (kN/m²)")
plt.title("Slab Load Components")

plt.savefig("outputs/engineering_plot.png")

plt.show()
```

## Engineering Conclusion

The reinforced concrete slab self-weight was calculated as
4.0 kN/m².

The total dead load was found to be 5.8 kN/m², while the
total service load reached 8.8 kN/m².

Applying the load combination 1.4D + 1.6L resulted in a
factored load of 12.92 kN/m².

The slab self-weight represents the largest dead-load
component and therefore has the greatest influence on the
overall slab loading. The calculated factored load should
be used for subsequent structural design and reinforcement
calculations.
